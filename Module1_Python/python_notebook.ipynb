#### With conditionals

```python
result = input("is it raining(type yes or no): ")

if result == "yes":
    print("take an umbrella")
else:
    print("enjoy your day")
```

```python
age = int(input("enter your age: "))

if age >= 18:
    print("you can vote")
else:
    print("you can not vote")
```

```python
age = int(input("enter your age: "))

if age >= 18 and age <= 100:
    print("you can vote")
elif age > 100:
    print("Age entered is not valid")
elif age < 0:
    print("Age entered is not valid and can not be negative")
else:
    print("you can not vote")
```

```python
num = int(input("enter num: "))

if num % 2 != 0: 
    print("num is odd")
else: 
    print("num is even")
```

```python
## guessing game

while True:
    num = int(input("Guess a number between 1 to 10: "))

    if num == 5:
        print("You guessed right")
        break
    elif num < 5:
        print("You guessed too low")
    else:
        print("You guessed too high")
```

```python
def is_even(num):
    if num % 2 == 0:
        print("num is even")
    else:
        print("num is odd")
```

```python
result = is_even(12)
```

```python
print(result)
```

```python
def is_odd(num):
    if num % 2 != 0:
        return "num is odd"
    else:
        return "num is even"
```

```python
print(is_odd(3))
```

```python
result = is_odd(10)
```

```python
print(result)
```
